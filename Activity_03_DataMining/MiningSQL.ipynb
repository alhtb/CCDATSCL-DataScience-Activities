#### Mine for association rules in the Movies dataset with three items on the left-hand side. Find support and confidence thresholds (need not be the same) so the number of association rules is more than 10 but less than 20.


```python
support = 0.03
confidence = 0.6
print(f"Support: {support}, Confidence: {confidence}")
```

```python
%%sql
drop table if exists Frequent;

create table Frequent as
select T1.item as item1, T2.item as item2, T3.item as item3, count(*) as lhs_count
from Movies T1, Movies T2, Movies T3
where T1.TID = T2.TID and T2.TID = T3.TID
and T1.item < T2.item and T2.item < T3.item
group by T1.item, T2.item, T3.item
having 1.0*count() / (select count(distinct TID) from Movies) > :support
```

```python
%%sql
-- This finds the rules and calculates confidence
select F.item1, F.item2, F.item3, T4.item as rhs,
       1.0*count() / F.lhs_count as confidence
from Frequent F, Movies T1, Movies T2, Movies T3, Movies T4
where F.item1 = T1.item
and F.item2 = T2.item and T1.TID = T2.TID
and F.item3 = T3.item and T2.TID = T3.TID
and T3.TID = T4.TID
and T4.item not in (F.item1, F.item2, F.item3)
group by F.item1, F.item2, F.item3, T4.item, F.lhs_count
having 1.0*count() / F.lhs_count > :confidence;

-- Count Check
select count(*) as Number_of_Confidence_Rules
from (
    select F.item1, F.item2, F.item3, T4.item
    from Frequent F, Movies T1, Movies T2, Movies T3, Movies T4
    where F.item1 = T1.item
    and F.item2 = T2.item and T1.TID = T2.TID
    and F.item3 = T3.item and T2.TID = T3.TID
    and T3.TID = T4.TID
    and T4.item not in (F.item1, F.item2, F.item3)
    group by F.item1, F.item2, F.item3, T4.item, F.lhs_count
    having 1.0*count() / F.lhs_count > :confidence
) as subquery;
```

#### Mine for association rules in the Movies dataset with three items on the left-hand side. Find support and lift thresholds so the number of association rules is more than 10 but less than 20.


```python
support = 0.05
liftthresh = 18.0
print(f"Support: {support}, Lift Threshold: {liftthresh}")
```

```python
%%sql
drop table if exists ItemFreq;

create table ItemFreq as
select item, count(*) as item_count
from Movies
group by item
```

```python
%%sql
drop table if exists Frequent;

create table Frequent as
select T1.item as item1, T2.item as item2, T3.item as item3, count(*) as lhs_count
from Movies T1, Movies T2, Movies T3
where T1.TID = T2.TID and T2.TID = T3.TID
and T1.item < T2.item and T2.item < T3.item
group by T1.item, T2.item, T3.item
having 1.0*count() / (select count(distinct TID) from Movies) > :support
```

```python
%%sql
-- This finds the rules and calculates lift
select F.item1, F.item2, F.item3, T4.item as rhs,
       (1.0*count() / F.lhs_count) / (1.0*IFq.item_count / (select count(distinct TID) from Movies)) as lift
from Frequent F, Movies T1, Movies T2, Movies T3, Movies T4, ItemFreq IFq
where F.item1 = T1.item
and F.item2 = T2.item and T1.TID = T2.TID
and F.item3 = T3.item and T2.TID = T3.TID
and T3.TID = T4.TID
and T4.item = IFq.item
and T4.item not in (F.item1, F.item2, F.item3)
group by F.item1, F.item2, F.item3, T4.item, F.lhs_count, IFq.item_count
having (1.0*count() / F.lhs_count) / (1.0*IFq.item_count / (select count(distinct TID) from Movies)) > :liftthresh;

-- Count Check
select count(*) as Number_of_Lift_Rules
from (
    select F.item1, F.item2, F.item3, T4.item
    from Frequent F, Movies T1, Movies T2, Movies T3, Movies T4, ItemFreq IFq
    where F.item1 = T1.item
    and F.item2 = T2.item and T1.TID = T2.TID
    and F.item3 = T3.item and T2.TID = T3.TID
    and T3.TID = T4.TID
    and T4.item = IFq.item
    and T4.item not in (F.item1, F.item2, F.item3)
    group by F.item1, F.item2, F.item3, T4.item, F.lhs_count, IFq.item_count
    having (1.0*count() / F.lhs_count) / (1.0*IFq.item_count / (select count(distinct TID) from Movies)) > :liftthresh
) as subquery;
```
